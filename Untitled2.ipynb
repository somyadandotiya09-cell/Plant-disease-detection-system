```python
import tensorflow as tf
import json
import os

# Load the 15-class dataset
train_ds = tf.keras.utils.image_dataset_from_directory(
    root,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=(128, 128),
    batch_size=32
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    root,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(128, 128),
    batch_size=32
)

class_names = train_ds.class_names
num_classes = len(class_names)

# Build a 15-class CNN model
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(128, 128, 3)),
    tf.keras.layers.Rescaling(1./255),

    tf.keras.layers.Conv2D(32, 3, activation='relu'),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(64, 3, activation='relu'),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(128, 3, activation='relu'),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(num_classes, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train model
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)

# Save model and disease names
model.save('/content/plant_disease_model.keras')

with open('/content/class_names.json', 'w') as f:
    json.dump(class_names, f)

print("15-class model trained and saved successfully!")
```

```python
import os
import json
import gradio as gr
import numpy as np
import tensorflow as tf

from pathlib import Path
from PIL import Image
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# 1. Connect Google Drive
from google.colab import drive
drive.mount('/content/drive')

# 2. Find your saved model and class names
drive_path = Path('/content/drive/MyDrive')

model_files = list(drive_path.rglob('plant_disease_model.keras'))
class_files = list(drive_path.rglob('class_names.json'))

if not model_files or not class_files:
    raise FileNotFoundError(
        "Could not find model or class_names.json in Google Drive."
    )

# 3. Load existing trained model
model = tf.keras.models.load_model(str(model_files[0]))

with open(class_files[0], 'r') as f:
    class_names = json.load(f)

if isinstance(class_names, dict):
    class_names = class_names["class_names"]

print("Model loaded successfully!")
print("Total disease classes:", len(class_names))


# 4. Treatment and prevention recommendations
def get_recommendations(disease):

    name = disease.lower().replace("_", " ")

    if "healthy" in name:
        return (
            "Your plant appears healthy!",
            "Continue regular watering, provide sufficient sunlight, "
            "use balanced fertilizer and inspect leaves regularly."
        )

    if "yellow leaf curl" in name or "mosaic virus" in name:
        return (
            "Remove severely infected plants. Control whiteflies and "
            "other insect vectors using appropriate locally approved methods.",
            "Use healthy seedlings, control weeds, monitor insects and "
            "remove infected plants promptly."
        )

    if "spider mite" in name:
        return (
            "Wash affected leaves with water. Use insecticidal soap "
            "according to its product label if needed.",
            "Inspect the undersides of leaves, avoid dusty conditions "
            "and maintain proper plant watering."
        )

    if "bacterial" in name:
        return (
            "Remove infected leaves and avoid working with wet plants. "
            "Seek locally recommended disease management advice.",
            "Use disease-free seeds, clean gardening tools, avoid "
            "overhead watering and rotate crops."
        )

    if "late blight" in name:
        return (
            "Remove infected leaves and plant debris. Seek local advice "
            "about approved late-blight management products.",
            "Avoid wet foliage, provide good air circulation and "
            "inspect plants regularly."
        )

    if "fungal" in name or any(x in name for x in [
        "early blight", "leaf mold", "septoria", "rust",
        "scab", "black rot", "esca", "leaf spot",
        "target spot", "leaf blight", "gray leaf spot"
    ]):
        return (
            "Remove affected leaves, dispose of infected plant debris "
            "and use locally approved fungicide only when appropriate.",
            "Avoid overhead watering, improve air circulation, "
            "keep tools clean and rotate crops."
        )

    return (
        "Remove badly affected leaves and seek local agricultural "
        "advice for the identified disease.",
        "Use healthy planting material, maintain proper spacing, "
        "avoid excess watering and inspect plants regularly."
    )


# 5. Disease prediction
def predict_disease(image):

    if image is None:
        return "Please upload a plant image."

    try:
        image = image.convert("RGB")
        image = image.resize((224, 224))

        image_array = np.array(image).astype("float32")
        image_array = np.expand_dims(image_array, axis=0)

        # Prediction
        predictions = model.predict(image_array, verbose=0)[0]

        index = np.argmax(predictions)
        disease = class_names[index]
        confidence = float(predictions[index]) * 100

        treatment, prevention = get_recommendations(disease)

        return f"""
## 🌿 Detection Result

*Disease:* {disease}

*Confidence:* {confidence:.2f}%

### 💊 Recommended Treatment
{treatment}

### 🛡️ Prevention
{prevention}
"""

    except Exception as e:
        return f"Prediction Error: {type(e)._name_}: {str(e)}"


# 6. Create website
demo = gr.Interface(
    fn=predict_disease,
    inputs=gr.Image(
        type="pil",
        label="Upload Plant Leaf Image"
    ),
    outputs=gr.Markdown(label="Detection and Recommendations"),
    title="🌱 Plant Disease Detection System",
    description=(
        "Upload a plant leaf image to identify diseases "
        "and receive treatment and prevention recommendations."
    ),
    theme="soft"
)

# 7. Launch website
demo.launch(share=True)
```

```python
import gradio as gr
import numpy as np
import tensorflow as tf
from PIL import Image

# Reuse your existing trained model
# If model is already loaded, it will not be loaded again.
try:
    model
except NameError:
    raise Exception("Your saved model must be loaded first.")

# Reuse your existing class names
try:
    class_names
except NameError:
    raise Exception("Your disease class names must be loaded first.")

# Treatment and prevention information
def get_advice(disease):

    d = str(disease).lower()

    if "healthy" in d:
        return (
            "Your plant appears healthy. Continue regular care.",
            "Provide sunlight, proper watering and regular inspection."
        )

    if "late blight" in d:
        return (
            "Remove infected leaves and seek locally approved disease management advice.",
            "Avoid overhead watering and improve air circulation."
        )

    if "early blight" in d:
        return (
            "Remove affected leaves and manage infected plant debris.",
            "Rotate crops, water near the soil and maintain plant spacing."
        )

    if "yellow leaf curl" in d or "mosaic" in d:
        return (
            "Remove severely infected plants and control insect vectors.",
            "Use healthy seedlings and monitor whiteflies regularly."
        )

    if "bacterial" in d:
        return (
            "Remove affected plant parts and clean gardening tools.",
            "Use disease-free seeds and avoid wet foliage."
        )

    if any(x in d for x in [
        "leaf mold", "septoria", "rust", "powdery mildew",
        "black spot", "target spot", "leaf spot", "fungal"
    ]):
        return (
            "Remove affected leaves and seek locally recommended treatment.",
            "Improve air circulation, avoid overhead watering and clean plant debris."
        )

    return (
        "Remove severely affected leaves and consult a local agricultural expert.",
        "Use healthy planting material, proper spacing and regular plant inspection."
    )


# Prediction function
def predict_disease(image):

    if image is None:
        return "Please upload a plant image."

    try:
        image = image.convert("RGB")

        input_shape = model.input_shape
        height = input_shape[1] or 224
        width = input_shape[2] or 224

        image = image.resize((width, height))

        arr = np.array(image).astype("float32")
        arr = np.expand_dims(arr, axis=0)

        # Assumes training images were rescaled to 0-1.
        # Change this only if your training preprocessing differed.
        arr = arr / 255.0

        predictions = model.predict(arr, verbose=0)[0]

        if len(predictions) != len(class_names):
            return "Error: Model output and disease class count do not match."

        index = int(np.argmax(predictions))
        disease = class_names[index]
        confidence = float(predictions[index]) * 100

        treatment, prevention = get_advice(disease)

        return f"""
# 🌿 Detection Result

### Disease: {disease}

### Confidence: {confidence:.2f}%

---

## 💊 Recommended Treatment

{treatment}

## 🛡️ Prevention Methods

{prevention}

---

🌱 Keep checking your plants regularly.
"""

    except Exception as e:
        return f"Detection error: {type(e)._name_}: {e}"


# Create complete website
demo = gr.Interface(
    fn=predict_disease,
    inputs=gr.Image(type="pil", label="Upload Plant Leaf"),
    outputs=gr.Markdown(),
    title="🌱 Plant Disease Detection System",
    description="Upload a leaf image to identify plant diseases and receive treatment and prevention advice.",
    theme="soft"
)

print("Website interface created successfully!")

# Launch website
demo.launch(share=True)
```
